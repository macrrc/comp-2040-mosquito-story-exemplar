# First Look: Winnipeg Weather at The Forks

The related dataset: Environment and Climate Change Canada's daily data for the WINNIPEG THE FORKS station, one CSV per year, as downloaded.

```python
from pathlib import Path
import pandas as pd

files = sorted(Path("../data/raw/weather").glob("*.csv"))
weather = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
print(len(files), "files,", weather.shape)
weather[["Date/Time", "Station Name", "Mean Temp (°C)", "Total Precip (mm)", "Total Precip Flag"]].head()
```

**One row is a day.** Checked: no date appears twice.

```python
weather.duplicated(subset=["Date/Time"]).sum()
```

## Coverage and gaps

```python
weather["date"] = pd.to_datetime(weather["Date/Time"])
summer = weather[weather["date"].dt.month.between(5, 9)]
print("years:", weather["date"].dt.year.min(), "to", weather["date"].dt.year.max())
print("summer days with no precipitation value:", summer["Total Precip (mm)"].isna().sum(), "of", len(summer))
summer.loc[summer["Total Precip (mm)"].isna(), "Total Precip Flag"].value_counts(dropna=False)
```
